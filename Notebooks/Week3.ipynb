```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("online_retail_II.csv")
```

```python
df["TotalAmount"] = df["Quantity"] * df["Price"]

AOV = df.groupby("Customer ID")["TotalAmount"].mean().reset_index()
AOV.columns = ["Customer ID", "AverageOrderValue"]

AOV.head()
```

```python
purchase_frequency = df.groupby("Customer ID")["Invoice"].nunique().reset_index()
purchase_frequency.columns = ["Customer ID", "PurchaseFrequency"]

purchase_frequency.head()
```

```python
cltv = pd.merge(AOV, purchase_frequency, on="Customer ID")

cltv["CLTV"] = cltv["AverageOrderValue"] * cltv["PurchaseFrequency"]

cltv.head()
```

```python
cltv["CustomerSegment"] = pd.qcut(
    cltv["CLTV"],
    q=3,
    labels=["Low Value", "Medium Value", "High Value"]
)

cltv.head()
```

```python
cltv.to_csv("Customer_CLTV.csv", index=False)

cltv.describe()
```
