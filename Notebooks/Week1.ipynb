```python
df["InvoiceMonth"] = df["InvoiceDate"].dt.to_period("M")
df["CohortMonth"] = df.groupby("Customer ID")["InvoiceMonth"].transform("min")
df[["Customer ID","InvoiceMonth","CohortMonth"]].head()
```

```python
invoice_year = df["InvoiceMonth"].dt.year
invoice_month = df["InvoiceMonth"].dt.month

cohort_year = df["CohortMonth"].dt.year
cohort_month = df["CohortMonth"].dt.month

df["CohortIndex"] = (
    (invoice_year - cohort_year) * 12
    + (invoice_month - cohort_month)
    + 1
)

df[["Customer ID","CohortMonth","InvoiceMonth","CohortIndex"]].head()
```

```python
cohort_data = df.groupby(
    ["CohortMonth", "CohortIndex"]
)["Customer ID"].nunique().reset_index()

cohort_matrix = cohort_data.pivot(
    index="CohortMonth",
    columns="CohortIndex",
    values="Customer ID"
)

cohort_matrix
```

```python
retention = cohort_matrix.divide(
    cohort_matrix.iloc[:,0],
    axis=0
) * 100

retention.round(2)
```

```python
retention.to_csv("RetentionMatrix.csv")
retention.head()
```
